inheritance:inherit properties of one class to child class

```python
class parent:
    def test_parent(self):
        print("this is my parent class")
```

```python
class child(parent):
    pass
```

```python
obj_child=child()
```

```python
obj_child.test_parent()
```

```python
class class1:
    def test_class1(self):
        print("this is my class")
```

```python
class class2(class1):
    def test_class2(self):
        print("this is my class")
```

```python
class class3(class2):
    def test_class3(self):
        print("this is my class")
```

```python
obj_class2=class3()
```

```python
obj_class2.test_class1()
```

```python
class classa:
    def test_classa(self):
        print("this is my class")
```

```python
class classb:
    def test_classb(self):
        print("this is my class")
```

```python
class classc(classa,classb):pass
```

```python
obj_classc=classc()
```

```python
obj_classc.test_classa()
```
