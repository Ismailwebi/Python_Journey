```python
class data_science:
    def syllabus(self):
        print("this is syllabus of data science");
```

```python
class web_dev:
    def syllabus(self):
        print("this is syllabus of web dev")
```

```python
def parcer(class_obj):
    for i in class_obj:
        i.syllabus()
```

```python
obj_data_science=data_science()
```

```python
obj_web_dev=web_dev()
```

```python
class_obj=[obj_data_science,obj_web_dev]
```

```python
parcer(class_obj)
```
