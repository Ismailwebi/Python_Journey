```python
class pwskill:
    def student_detail(self,name,email):
        print(name,email)
    @staticmethod
    def details(mentor_list):
        print(mentor_list)
```

```python
pwskill.details(["ismail",'ismail'])
```

```python
pw=pwskill()
```

```python
pw.student_detail("ismail","ismail")
```
